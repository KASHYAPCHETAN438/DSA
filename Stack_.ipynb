# STACK – Data Structure

## 🔹 What is a Stack?

- A Stack is a linear data structure that follows the principle of:

- LIFO (Last In, First Out)

    - The element inserted last is removed first

- Example:

    - Stack of plates, Undo operation

<br>


##🔹 Basic Terminology

- Push → Insert an element

- Pop → Remove an element

- Peek / Top → View top element

- Overflow → Stack is full

- Underflow → Stack is empty

<br>


## 🔹  Representation

- A stack can be implemented using:

    - Array

    - Linked List

## 🔹 Stack Implementation Using Array

```python
stack = []
max_size = 5

def push(x):
    if len(stack) == max_size:
        print("Overflow")
    else:
        stack.append(x)

def pop():
    if not stack:
        print("Underflow")
    else:
        stack.pop()

def peek():
    if not stack:
        print("Empty Stack")
    else:
        print(stack[-1])
```

#  Stack Operations

## 1️⃣ Push Operation (Insertion)

- Adds an element at the top of the stack.

#### Algorithm:

- Check if stack is full

- Increase top

- Insert element at top

#### Key Points:

- Insertion happens only at top

- Time Complexity: O(1)

<br>

```python
PUSH(stack, value):
    if top == max - 1:
        print "Overflow"
    else:
        top = top + 1
        stack[top] = value

```

## 2️⃣ Pop Operation (Deletion)

- Removes the top element from the stack.

#### Algorithm:

- Check if stack is empty
    - top == -1

- Remove element at top
   

- Decrease top

  - top = top - 1


#### Key Points:

- Deletion happens only at top

- Time Complexity: O(1)

```python
POP(stack):
    if top == -1:
        print "Underflow"
    else:
        top = top - 1
```

## 3️⃣ Peek / Top Operation
- Returns the top element without removing it.

### Algorithm:


```python
PEEK(stack):
    if top == -1:
        print "Stack Empty"
    else:
        return stack[top]
```

## 4️⃣ isEmpty Operation

- Checks whether stack is empty.

```python
isEmpty():
    return top == -1
```

## 5️⃣ isFull Operation

- Checks whether stack is full.

```python
isFull():
    return top == max - 1
```
