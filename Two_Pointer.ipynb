# What is Two Pointer Technique in DSA ?

- The Two Pointer technique is a problem-solving approach where two pointers (indexes) are used to traverse a data structure (mostly arrays or strings) to reduce time complexity.

- Instead of using nested loops (O(n²)), we use two pointers to solve the problem in O(n) time.

<br>

### Why Two Pointer Technique is Used?

- Improves time efficiency

- Reduces nested loops

- Easy to implement

- Frequently asked in exams & interviews

<br>


# When to Use Two Pointer Technique?

- Array or string involved
- Sorted data
- Need optimized solution
- Brute force is too slow




<br>

# Check if pair sum exists or not

```python
a=[1,2,3,4,5,6,8,9,10,43]

l=0
r=len(a)-1

t=11

while l<r:

    cr = a[l] + a[r]

    if cr == t:

        print("Pair found : ",l, r)
        print("Value of the Pair : ",a[l], a[r])
        print("Tager sum :" , a[l]+a[r],"\n\n")


        l += 1
        r -= 1

    elif t < cr:
        r -= 1

    else:
        l += 1
```

# Palindrom Check

```python
s = "abba"

l = 0
r = len(s) - 1

while l < r:
    if s[l] != s[r]:
        print("Given String is not Palindrome")
        break

    l += 1
    r -= 1

else:
    print("Given String is Palindrome")
```

# Reverse a String

```python
s = "helloworld"

def reverseString(s):

    s = list(s)

    l = 0
    r = len(s) - 1

    while l < r:
        s[l], s[r] = s[r], s[l]

        l += 1
        r -= 1

    return "".join(s)


ans = reverseString(s)

print(ans)
```

```python
class Solution(object):

    def hasCycle(self, head):

        # slow pointer -> 1 step move karega
        slow = head

        # fast pointer -> 2 step move karega
        fast = head

        # jab tak fast aur fast.next exist kare
        while fast and fast.next:

            # slow 1 step
            slow = slow.next

            # fast 2 step
            fast = fast.next.next

            # agar dono same node pe aa gaye
            # to cycle present hai
            if slow == fast:
                return True

        # agar fast NULL ho gaya
        # matlab cycle nahi hai
        return False
```

# What is Sliding Window Technique in DSA?

- The Sliding Window technique is an optimization method used to solve problems involving subarrays or substrings by maintaining a window (range) that slides over the data.

- Instead of recalculating again and again, we reuse previous computations, which reduces time complexity from O(n²) → O(n).



# 🔁 Sliding Window = Window + Slide

- Window → current subarray

- Slide → window ko aage move karna

<br>

# Types of Sliding Window


##1️⃣ Fixed Size Sliding Window

👉 Window size is constant (k)

Used when:
“subarray of size K” is mentioned.

📌 Examples:

Maximum sum subarray of size K

First negative number in every window of size K



```python
arr = [2, 1, 5, 1, 3]
k = 3

for i in range(len(arr) - k + 1):
    window = arr[i:i+k]
    print(window)
```

## 2️⃣ Variable Size Sliding Window

👉 Window size expands and shrinks

Used when:
“longest / shortest subarray or substring” is mentioned.

📌 Examples:

Longest substring without repeating characters

Smallest subarray with sum ≥ K

```python
# Print subarrays whose sum is ≤ 5


arr = [1, 2, 1, 1, 3]

left = 0
current_sum = 0

for right in range(len(arr)):
    current_sum += arr[right]

    while current_sum > 5:
        current_sum -= arr[left]
        left += 1

    print(arr[left:right+1])
```
