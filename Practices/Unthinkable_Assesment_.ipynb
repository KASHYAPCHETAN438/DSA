```python
# QUESTION .1
# Question on binary number. The input is n binary numbers in n different rows, find the unrepeated binary row (2020).





def uniqueBinaryNumber(lst):
    # Convert binary strings to integers
    nums = [int(b, 2) for b in lst]

    ans = 0
    for i in nums:
        ans = ans ^ i   # XOR to cancel out pairs

    # Convert back to binary string
    return bin(ans)[2:]   # [2:] se '0b' remove hoga

# Example
lst= ["101", "111", "101", "110", "110"]


print(uniqueBinaryNumber(lst))
```

```python
# QUESTION . 3

lst = [1, 4, 8, 5, 2, 7, 5, 2, 4, 6]

n = int(input("Enter the sum: "))

count = 0
pairs = []

for i in range(len(lst)):
    for j in range(i+1, len(lst)):   # i+1 se start → no duplicates
        if lst[i] + lst[j] == n:
            print(f"Found pair at index ({i},{j}) : values ({lst[i]},{lst[j]})")
            pairs.append((lst[i], lst[j]))
            count += 1

print(f"\nTotal pairs = {count}")
print("All pairs:", pairs)
```

```python
# QUESTION . 6


# Letter Balance



s = "aabbcc"
unique_chars = set(s)  # unique characters ka set banalo
counts = []

for ch in unique_chars:
    counts.append(s.count(ch))  # har character ka count add karo

# Check if all counts same hain
if len(set(counts)) == 1:
    print("Balanced")
else:
    print("Not Balanced")
```

```python
# QUESTION . 6

# Equal Frequency Balance


s = "RLRRLL"
count = 0
r_count = 0
l_count = 0

for char in s:
    if char == 'R':
        r_count += 1
    elif char == 'L':
        l_count += 1
    if r_count == l_count:
        count += 1

print("Balanced substrings:", count)
```

```python
# QUESTION . 6

# Parentheses / Brackets Balance




def isBalanced(s):
    stack = []
    for char in s:
        if char == '(':
            stack.append(char)
        elif char == ')':
            if not stack:
                return False
            stack.pop()
    return len(stack) == 0

print(isBalanced("(()())"))  # True
print(isBalanced("(()"))     # False

```

# Problem Statement

* You are playing a game of "First Word" with your family.
- Here, each player gets an array S of length N
- The game's motive is to get the minimum points by rearranging S using the following operation any number of times:

- You can select one of the first K elements from S, remove it from there, and append it to the end of S.

**After rearranging the array S,**

 - you need to calculate the points using this formula:

image.png


**Where:**

- Position(ch) = index of character in English alphabet (a=1, b=2, ..., z=26).

- (N-i) = weight of position (higher for elements at start, lower at end).


**Constraints**

- 1 ≤ N ≤ 1000
- 1 ≤ K ≤ N
- S contains lowercase English letters


**Your Task**

- Print the minimum points achievable.

**Input Format**

- First line: An integer N (length of array).

- Second line: An integer K (allowed prefix size for rotation).

- Third line: An array S of string

**Output Format**

- A single integer → minimum number of points possible.

**Input**
- 4
- 2
- hack



**Process:**

- K=2 → we can move h or a to the end.

- Different rotations:

- hack → hcka → hkac → kach → achk

- Calculate points for each.

- Points("kach") = 4**1 + 3**3 + 2**8 + 1**11 = 40

```python
def position(ch):
    return ord(ch) - ord('a') + 1

def calculate_points(s):
    n = len(s)
    total = 0
    for i in range(n):
        total += position(s[i]) * (n - i)
    return total

def min_points(N, K, S):
    min_score = calculate_points(S)

    # Try all rotations up to K
    for i in range(1, K + 1):
        rotated = S[i:] + S[:i]
        score = calculate_points(rotated)
        min_score = min(min_score, score)

    return min_score

# ---------- Main ----------
if __name__ == "__main__":
    N = int(input().strip())
    K = int(input().strip())
    S = input().strip()

    result = min_points(N, K, S)
    print(result)
```

```python
def rotate_string(s, i):
    """
    Rotate string s by i positions to the left.
    Example: rotate_string("hack", 2) -> "ckha"
    """
    i = i % len(s)  # in case i > length of string
    return s[i:] + s[:i]

# ---------- Example ----------
s = "hack"
k = 2

rotated1 = rotate_string(s, 1)  # rotate by 1
rotated2 = rotate_string(s, 2)  # rotate by 2

print("Original:", s)
print("Rotate by 1:", rotated1)
print("Rotate by 2:", rotated2)
```
