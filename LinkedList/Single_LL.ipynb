

# Linked List

- A **Linked List** is a dynamic data structure ideal for fast insertions and deletions. Unlike arrays, its elements aren't stored contiguously but are linked via pointers.

<br>

### Anatomy of a Linked List

- A Liked list is Collection of nodes. there are two types of nodes are existed.

    1) **single node** ----> data, next
    2) **double node** ----> prev, data, next


- Each  **nodes** having:
    - **Data**: The stored value.
    - **Next Pointer**: A reference to the next node.

- The list starts with a **Head** node and ends with a node having a **null** Next pointer.

<br>

### Key Features

- **Dynamic Size**: Adapts to data volume.
- **Non-Contiguous Memory**: Flexibility in storage.
- **Fast Insertions/Deletions**: Limited pointer adjustments needed.

## **Types of LinkedList with Real-Life Examples**



| LinkedList Type                | Meaning                                     | Real-Life Example                                                     | Why?                                 |
| ------------------------------ | ------------------------------------------- | --------------------------------------------------------------------- | ------------------------------------ |
| **Singly LinkedList**          | Each node links to **next** node only       | **Music Playlist – Play Next Song**                                   | You can only go forward song-by-song |
| **Doubly LinkedList**          | Each node links to both **next & previous** | **Browser Back & Forward Buttons**                                    | You can move in both directions      |
| **Circular LinkedList**        | Last node links back to **first** node      | **Round Robin Scheduling in OS / Carousel Menu**                      | It repeats in a continuous loop      |
| **Circular Doubly LinkedList** | Can move both sides **circularly**          | **Navigation in Image Galleries** (→ next, ← previous, infinite loop) | You can scroll left/right forever    |



<br>


1. **Singly Linked List**:
    
    - Each node has a single pointer to the next node.
    - Traversal is unidirectional: from head to tail.
    

   ![Singly Linked List](https://firebasestorage.googleapis.com/v0/b/dev-stack-app.appspot.com/o/linked-lists%2Fsingly-linked-list.svg?alt=media&token=c6e2ad4f-e2d4-4977-a215-6253e71b6040)

2. **Doubly Linked List**:
    - Each node have two pointers: one pointing to the next node and another to the previous node.
    - This allows for bidirectional traversal.

    ![Doubly Linked List](https://firebasestorage.googleapis.com/v0/b/dev-stack-app.appspot.com/o/linked-lists%2Fdoubly-linked-list.svg?alt=media&token=5e14dad3-c42a-43aa-99ff-940ab1d9cc3d)

3. **Circular Linked List**:
    - Like a singly linked list, but the tail node points back to the head, forming a loop.
![Circular Linked List](https://firebasestorage.googleapis.com/v0/b/dev-stack-app.appspot.com/o/linked-lists%2Fcircular-linked-list.svg?alt=media&token=b3b96bc7-3b16-4d07-978f-e4774a048ee1)

4. **Multi-level Linked List**:
     - This specialized type has nodes with multiple pointers, each pointing to different nodes.
     - It's often used in advanced data structures like multi-level caches.

   ![Multi-level Linked List](https://firebasestorage.googleapis.com/v0/b/dev-stack-app.appspot.com/o/linked-lists%2Fmulti-level-linked-list.svg?alt=media&token=967af5cf-8a95-4c05-a8fe-fb70f2b7ea57)

### **Operation and Time Complexity**

| **Operation**                      | **Description**                  | **Time Complexity** |
| ---------------------------------- | -------------------------------- | ------------------- |
| **Traversal**                      | Visit all nodes                  | O(n)                |
| **Insertion (beginning)**          | Add node at start                | O(1)                |
| **Insertion/Deletion (elsewhere)** | Add/remove node at any position  | O(n)                |
| **Search**                         | Find a specific node             | O(n)                |
| **Sorting**                        | Arrange nodes (e.g., merge sort) | O(n log n)          |
| **Merging**                        | Combine two linked lists         | O(n)                |
| **Reversal**                       | Reverse node order               | O(n)                |

### Insertion at a Given Position (Middle Insertion)

```python
def insertionAtGivenPosition(head, pos, data):
    new_node = node(data)

    # Case 1: Insert at first position
    if pos == 1:
        new_node.next = head
        head = new_node
        return head

    temp = head
    count = 1

    # Traverse to (pos-1) node
    while temp is not None and count < pos - 1:
        temp = temp.next
        count += 1

    # If position is invalid (greater than list size)
    if temp is None:
        print("Position out of range")
        return head

    # Insert new node
    new_node.next = temp.next
    temp.next = new_node

    return head


print("--------Insert at given position---------\n")
head=insertionAtGivenPosition(head, 6, 14)
printlist(head)
```

## Insertion at the End (Tail Insertion)

```python

def insertionAtEnd(head, data):
    new_node = node(data)

    # If list is empty
    if head is None:
        head = new_node
        return head

    # Otherwise traverse till the last node
    temp = head
    while temp.next is not None:
        temp = temp.next

    # Link last node to new node
    temp.next = new_node

    return head



print("Already we have this linked list:")
printlist(head)

print("\n--------Insert at End---------\n")
head = insertionAtEnd(head, 101)
printlist(head)
```

 ## Delete at given location


#### # internall working --> https://www.notion.so/Single-Linked-List-289b4abf5f158040b64ac726e74852c4?source=copy_link

```python
#[10] → [20] → [30] → [40 | next] → [50 | next] → [60] → [70] → [80] → [90]
#                               ↑
#                             prev
#                                           ↑
#                                           └── temp





def deleteAtGivenPos(head, pos):
    if head is None:
        print("List is empty!")
        return head

    temp = head

    # If deleting the first node
    if pos == 1:
        head = temp.next            #    temp=temp.next
        temp = None
        return head

    # Traverse to the node before the one to delete
    count = 1
    prev = None

    while temp is not None and count < pos:
        prev = temp          # prev --> pos-1
        temp = temp.next     # temp-->pos
        count += 1

    # If position is invalid
    if temp is None:
        print("Position out of range!")
        return head

    # Adjust links
    prev.next = temp.next     # temp.next-->pos+1
    temp = None
    return head


print("Before deletion linked list is :")
printlist(head)

print("\n--------Deletion at the given position---------\n")
head = deleteAtGivenPos(head,5)
printlist(head)
```

 # Delete at last


```python
def deleteAtEnd(head):
    # Case 1: Empty list
    if head is None:
        print("List is empty!")
        return head

    # Case 2: Only one node
    if head.next is None:
        head = None
        return head

    # Case 3: More than one node
    temp = head
    while temp.next.next is not None:   # stop at 2nd last node
        temp = temp.next

    temp.next = None   # remove link to last node
    return head


print("Before deletion linked list is :")
printlist(head)

print("\n--------Deletion at the given position---------\n")
head = deleteAtEnd(head)
printlist(head)
```

# Delete elements

```python
def deleteByValue(head, value):
    # Case 1: If list is empty
    if head is None:
        print("List is empty")
        return head

    # Case 2: If head node itself holds the value
    if head.data == value:
        print(f"Deleting head node: {value}")
        head = head.next  # Move head to next node
        return head

    # Case 3: Search for the node to delete
    temp = head
    prev = None
    while temp is not None and temp.data != value:
        prev = temp
        temp = temp.next

    # If value not found
    if temp is None:
        print("Value not found in list")
        return head

    # Skip the node to delete
    print(f"Deleting node with value: {value}")
    prev.next = temp.next
    temp = None  # Free the deleted node
    return head


print("Before deletion linked list is :")
printlist(head)

print("\n--------Deletion at the given position---------\n")
head = deleteByValue(head, 10)
printlist(head)
```

# 6. Middle of element in linked list

```python
def middleElementLL(head):
    # Step 1: Count total nodes
    count = 0
    temp = head
    while temp is not None:
        count += 1
        temp = temp.next

    # Step 2: Find middle position
    mid_index = count // 2

    # Step 3: Traverse again to reach middle
    temp = head
    for i in range(mid_index):
        temp = temp.next

    # Step 4: Return middle node data
    return temp.data

print("-------Element in LL---------")
printlist(head)


print("\n----Middle Element of LL-----")
midEl=middleElementLL(head)
print(midEl)
```

## After indertion in DESC order

```python
def sortAfterInDESC(head):
    if head is None:
        return None

    temp1 = head
    while temp1 is not None:
        temp2 = temp1.next
        while temp2 is not None:
            if temp1.data < temp2.data:
                # Swap data for descending order
                temp1.data, temp2.data = temp2.data, temp1.data
            temp2 = temp2.next
        temp1 = temp1.next
    return head


print("\n------Before sorting:-------")
printlist(head)

head = sortAfterInDESC(head)

print("\n------After sorting (DESC):------")
printlist(head)
```

##  After indertion in ASC order

```python
def sortAfterInASC(head):
    if head is None:
        return None

    temp1 = head
    while temp1 is not None:
        temp2 = temp1.next
        while temp2 is not None:
            if temp1.data > temp2.data:
                # Swap the data
                temp1.data, temp2.data = temp2.data, temp1.data
            temp2 = temp2.next
        temp1 = temp1.next
    return head


print("------Before sorting:-------")
printlist(head)

head = sortAfterInASC(head)

print("\n\n------After sorting (ASC):------")
printlist(head)
```

##  Insertion in ASC order
  

```python
def insertionInASC(head, data):
    new_node = node(data)

    # If list is empty or new data is smaller than head
    if head is None or data < head.data:
        new_node.next = head
        head = new_node
        return head

    temp = head
    while temp.next is not None and temp.next.data < data:
        temp = temp.next

    new_node.next = temp.next
    temp.next = new_node
    return head


print("------Before Insertion (ASC):-------")
printlist(head)

head = insertionInASC(head, 25)

print("\n------After Insertion in ASC order:-------")
printlist(head)
```

##  Insertion in DESC order
  

```python
def insertionInDESC(head, data):
    new_node = node(data)

    # If list is empty or new data is greater than head
    if head is None or data > head.data:
        new_node.next = head
        head = new_node
        return head

    temp = head
    while temp.next is not None and temp.next.data > data:
        temp = temp.next

    new_node.next = temp.next
    temp.next = new_node
    return head


print("\n------Before Insertion (DESC):-------")
printlist(head)

head = insertionInDESC(head, 35)

print("\n------After Insertion in DESC order:-------")
printlist(head)
```

#9. Remove duplicated
 - From sorted linked list

## From sorted linked list

```python
def removeDuplicate(head):
    temp = head

    while temp is not None and temp.next is not None:
        if temp.data == temp.next.data:
            # Skip duplicate node
            temp.next = temp.next.next
        else:
            temp = temp.next

    return head

print("------Before removing duplicates:------")
printlist(head)

head = removeDuplicate(head)

print("\n------After removing duplicates:------")
printlist(head)
```

## From unsorted linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def printlst(head):
    temp = head
    while temp:
        print(temp.data, end=" -> ")
        temp = temp.next
    print("None")


# Create unsorted list with duplicates    10 → 20 → 20 → 14 → 30 → 30 → 40 → 14

head1 = Node(10)
head1.next = Node(20)
head1.next.next = Node(20)
head1.next.next.next = Node(14)
head1.next.next.next.next = Node(30)
head1.next.next.next.next.next = Node(30)
head1.next.next.next.next.next.next = Node(40)
head1.next.next.next.next.next.next.next = Node(14)


```

```python
def removeDuplicate(head):
    if head is None:
        return None

    seen = set()       # Store unique data values
    temp = head
    prev = None

    while temp is not None:
        if temp.data in seen:
            # Remove duplicate node
            prev.next = temp.next
        else:
            seen.add(temp.data)
            prev = temp
        temp = temp.next

    return head



print("------Before removing duplicates:------")
printlst(head1)

head = removeDuplicate(head1)

print("\n------After removing duplicates:------")
printlst(head1)
```

#10. Copy the list




```python
def reverseList(head):
    prev = None
    current = head

    while current is not None:
        next_node = current.next  # store next
        current.next = prev       # reverse pointer
        prev = current
        current = next_node

    head = prev
    return head


print("------Before Reverse------")
printlist(head)

print("\n------After Reverse------")
head = reverseList(head)
printlist(head)
```

#11. Reverse the linked list


```python
def reverseList(head):
    prev = None
    current = head

    while current is not None:
        next_node = current.next  # store next
        current.next = prev       # reverse pointer
        prev = current
        current = next_node

    head = prev
    return head


print("------Before Reverse------")
printlist(head)

print("\n------After Reverse------")
head = reverseList(head)
printlist(head)
```

#13. nth node from begining / ending
  - From Beginning
  - From End

##  From Beginning


```python
def nthFromBeginning(head, n):
    temp = head
    count = 1
    while temp is not None and count < n:
        temp = temp.next
        count += 1
    if temp is None:
        print("Position out of range")
    else:
        print(f"{n}th node from beginning is:", temp.data)


print("------Initil LL------")
printlist(head)

print("\n------ N th node from beginning ------\n")
nthFromBeginning(head, 3)
```

##  From End

```python
def nthFromEnd(head, n):
    first = second = head
    for _ in range(n):
        if first is None:
            print("Position out of range")
            return
        first = first.next
    while first is not None:
        first = first.next
        second = second.next
    print(f"{n}th node from end is:", second.data)


print("------Initil LL------")
printlist(head)

print("\n------ N th node from ending ------")
nthFromEnd(head, 4)
```

#14 paliandrom or not

```python
def isPalindrome(head):
    stack = []        # Create an empty stack
    temp = head

    # Step 1: Push all elements into stack
    while temp is not None:
        stack.append(temp.data)   # push current node's data into stack
        temp = temp.next          # move to next node

    # Step 2: Compare again from head
    temp = head
    while temp is not None:
        if temp.data != stack.pop():  # compare node data with last element in stack
            return False              # mismatch → not a palindrome
        temp = temp.next

    return True                        # all matched → palindrome


isPalindrome(head)
```
