## 09. Move Zeros to the End

### 📝 Problem
Move all zeros to the end while preserving the order of non-zero elements.

### Example

- lst = [0, 1, 0, 3, 12]


#### Expected Output

- [1, 3, 12, 0, 0]

#### Conditions
- Do not use sorting.
- Maintain the relative order of non-zero values.



image.png

````python
# Time Complexity O(n)

lst = [0, 1, 0, 3, 12]

j = 0

for i in range(len(lst)):
    if lst[i] != 0:
        lst[i], lst[j] = lst[j], lst[i]
        j += 1

print(lst)
````

````python
# Time Complexity O(n2)

lst = [0, 1, 0, 3, 12]

for i in range(len(lst)):

    if lst[i] == 0:

        for j in range(i + 1, len(lst)):

            if lst[j] != 0:
                lst[i], lst[j] = lst[j], lst[i]
                break

print(lst)
````

## 11. Find Common Elements

### 📝 Problem
Given two tuples, find elements that appear in both tuples.

#### Example

- t1 = (1, 2, 3, 4, 5)
- t2 = (4, 5, 6, 7, 8)


#### Expected Output

- Common elements: (4, 5)






````python
# Using Time Complexity O(n)

t1 = (1, 2, 3, 4, 5)
t2 = (4, 5, 6, 7, 8)

elements = set(t1)

for i in t2:
    if i in elements:
        print(f"Find Common element : {i}")
````

````python
# Using Time Complexity O(n2)

t1 = (1, 2, 3, 4, 5)
t2 = (4, 5, 6, 7, 8)

for i in t1:
    for j in t2:
        if i == j:
            print(f"Find Common element : {i}")
````


## 17. Invert a Dictionary

### 📝 Problem
Swap dictionary keys and values.

### Example
```python
data = {
    "a": 1,
    "b": 2,
    "c": 3
}
```

### Expected Output
```python
{
    1: "a",
    2: "b",
    3: "c"
}
```

### Assumption
- All values are unique.



````python
data = {
    "a": 1,
    "b": 2,
    "c": 3
}

print("Before Swap:")
print(data)

for key, value in list(data.items()):
    data[value] = key
    del data[key]

print("\nAfter Swap:")
print(data)
````

````python
data = {
    "a": 1,
    "b": 2,
    "c": 3
}

print("Dictionary Before Swap Key and Values")
print(data)

new_data = {}

for key in data:
    value = data[key]
    new_data[value] = key

print("\nDictionary After Swap Key and Values")
print(new_data)
````


## 20. Group Words by Their First Character

### 📝 Problem
Given a list of words, create a dictionary where each key is the first character and the value is a list of words starting with that character.

### Example

- words = ["apple", "ant", "ball", "bat", "cat"]


### Expected Output
```python
{
    "a": ["apple", "ant"],
    "b": ["ball", "bat"],
    "c": ["cat"]
}
```







````python
words = ["apple", "ant", "ball", "bat", "cat"]

empt = {}
keys = []

# Find unique first characters
for i in words:
    keys.append(i[0])

keys = list(set(keys))
keys.sort()

# Find values
for key in keys:
    empt[key] = []

    for word in words:
        if word[0] == key:
            empt[key].append(word)

print(empt)
````
