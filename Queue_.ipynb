##🔹 What is a Queue?

- A Queue is a linear data structure that follows the principle:

-  FIFO (First In, First Out)

    - The element inserted first is removed first

- Example:
    - Queue at a ticket counter

<br>

##🔹 Basic Terminology

- Enqueue → Insert an element

- Dequeue → Remove an element

- Front → First element

- Rear → Last element

- Overflow → Queue is full

- Underflow → Queue is empty

<br><br>

## Representation:
- A queue can be implemented using:

    - Array

    - Linked List

## 1️⃣ Enqueue Operation (Insertion)

- Adds an element at the rear of the queue.

#### Algorithm:

- Check if queue is full

- If full → Overflow

- If queue is empty, set front = 0

- Increase rear

- Insert element at rear


#### Key Points

- Insertion happens at rear

- Time Complexity: O(1)

<br>

<br>

## 2️⃣ Dequeue Operation (Deletion)
- Removes an element from the front of the queue.

#### Algorithm (Dequeue)

- Check if queue is empty

- If empty → Underflow

- Remove element at front

- Increase front

- If front > rear, reset both to -1


#### Key Points

- Deletion happens at front

- Time Complexity: O(1)

```python
DEQUEUE(queue):
    if front == -1 or front > rear:
        print "Underflow"
        return

    front = front + 1
```

<br>

## 3️⃣ Peek / Front Operation
- Returns the front element without removing it.

```python
PEEK(queue):
    if front == -1:
        print "Queue is Empty"
    else:
        return queue[front]
```

# 🔹 Queue Implementation Using Array

```python
queue = []
max_size = 5

def enqueue(x):
    if len(queue) == max_size:
        print("Overflow")
    else:
        queue.append(x)

def dequeue():
    if not queue:
        print("Underflow")
    else:
        queue.pop(0)

def peek():
    if not queue:
        print("Queue Empty")
    else:
        print(queue[0])
```

## 🔹 Advantages of Queue

- Maintains order
- Efficient insertion & deletion
- Useful in scheduling

<br>

## 🔹 Disadvantages of Queue

- No random access
- Fixed size (array implementation)
- Wastage of space in simple queue

<br>

## 🔹 Applications of Queue

- CPU scheduling

- Printer spooling

- Breadth First Search (BFS)

- Call center systems

- Data buffering
