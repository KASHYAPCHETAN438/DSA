```python
s ="abcei"

vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0
flipped = ""

for ch in s:
    if ch.isalpha():                      # Only alphabets
        if ch in vowels:
            vowel_count += 1
        else:
            consonant_count += 1

        # Flip case
        if ch.islower():
            flipped += ch.upper()
        else:
            flipped += ch.lower()
    else:
        continue   # Ignore other symbols

print(consonant_count, vowel_count, flipped)
```

# Type-1

```python
def isPerfectArray(arr):
    n = len(arr)

    i = 0

    # 1️⃣ Strictly increasing
    while i + 1 < n and arr[i] < arr[i + 1]:
        i += 1

    # 2️⃣ Constant
    while i + 1 < n and arr[i] == arr[i + 1]:
        i += 1

    # 3️⃣ Strictly decreasing
    while i + 1 < n and arr[i] > arr[i + 1]:
        i += 1

    # If we reached the end → perfect
    return i == n - 1


# ---------------------
# Test
arr1 = [1, 8, 8, 8, 3, 2]
arr2 = [1, 1, 2, 2, 1]

print("Yes" if isPerfectArray(arr1) else "No")
print("Yes" if isPerfectArray(arr2) else "No")
```

# Type-2 (Optimal way)

```python
s = '''In 1989, the world witnessed significant political changes, including the fall of the Berlin Wall. In 1989, the internet started to become more accessible to the public. By 2000, the Y2K bug was a major concern, but the transition into the new millennium went smoothly. In 2000, the global financial crisis affected economies worldwide and ends in 2010.'''


Year_feq = {}

for i in s.split():
    word = i.strip(",.")  # Remove punctuation

    if word.isdigit():  # Check if numeric
        year = int(word)
        if year in Year_feq:
            Year_feq[year] += 1
        else:
            Year_feq[year] = 1


print(Year_feq)

```

# Type-3




```python
s = '''In 1989, the world witnessed significant political changes, including the fall of the Berlin Wall.
In 1989, the internet started to become more accessible to the public.
By 2000, the Y2K bug was a major concern, but the transition into the new millennium went smoothly.
In 2000, the global financial crisis affected economies worldwide and ends in 2010.'''

l = s.split()
unique = {}

for i in l:
    # Remove punctuation like commas and periods
    word = i.strip(',.')

    # Check if it's a year (4 digits and numeric)
    if word.isdigit() and len(word) == 4:
        if word in unique:
            unique[word] += 1
        else:
            unique[word] = 1

for year, count in unique.items():
    print(f"{year}: {count}")
```

# Type-4  (using Regular expresion)

```python
import re

s = '''In 1989, the world witnessed significant political changes, including the fall of the Berlin Wall.
In 1989, the internet started to become more accessible to the public.
By 2000, the Y2K bug was a major concern, but the transition into the new millennium went smoothly.
In 2000, the global financial crisis affected economies worldwide and ends in 2010.'''

# Extract all 4-digit years using regex
years = re.findall(r'\b\d{4}\b', s)

# Count occurrences
year_count = {}
for y in years:
    year_count[y] = year_count.get(y, 0) + 1

# Print output
for year, count in year_count.items():
    print(f"{year}: {count}")
```
