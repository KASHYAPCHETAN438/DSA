# Doubly Linked List (DLL)


<br>

##🔹 What is a Doubly Linked List?

- A Doubly Linked List is a linear data structure where:

- Each node have two pointers: one pointing to the next node and another to the previous node.

- Because of prev and next, traversal is possible in both directions.

- This allows for bidirectional traversal.

##🔹 Node Structure

- Each node has 3 parts:

    - data → stores value

    - prev → points to previous node

    - next → points to next node


```python

class Node:
  def __init__(self, data):
    self.data = data
    self.prev = None
    self.next = None
```

## 🔸Forward Traversal
- orward traversal in a Doubly Linked List (DLL) means visiting nodes from the first node (head) to the last node (tail) using the next pointer.

#### ALgorithm :

- Start from the head node of the doubly linked list.

- If head == NULL, the list is empty → stop.

- Set a temporary pointer temp = head.

- While temp != NULL:

    - Visit / process temp.data

    - Move to the next node using temp = temp.next

- Stop when temp becomes NULL.

<br>

#### Time Complexity: O(n)


```python
def forward_traversal(head):
    temp = head
    while temp:
        print(temp.data, end=" <-> ")
        temp = temp.next
    print("None")
```

#### Algorithm:

- Start from the tail node of the doubly linked list.

- If tail == NULL, the list is empty → stop.

- Set a temporary pointer temp = tail.

##🔸 Insert at Beginning
- Insertion at the beginning in a Doubly Linked List (DLL) means adding a new node before the current head node, and making the new node the new head.


#### Algorithm:

- Create a new node with given data.
- Set newNode.next to current head.
- Set newNode.prev = NULL.
- If the list is not empty, set head.prev = newNode.
- Update head = newNode

#### Time Complexity: O(1)


```python
def insert_at_beginning(head, data):
    new_node = Node(data)
    new_node.next = head
    if head:
        head.prev = new_node
    head = new_node
    return head
```


##🔸 Insert at End
- Insertion at the end in a Doubly Linked List (DLL) means adding a new node after the last node (tail) of the list.

#### Algorithm:

- Create a new node with given data.
- If the list is empty (head == NULL):
    - Set newNode.prev = NULL

    - Set newNode.next = NULL

    - Make head = newNode

    - Stop

- Otherwise:

  - Traverse the list to reach the last node

- Set last.next = newNode

- Set newNode.prev = last

- Set newNode.next = NULL

#### Time Complexity: O(n) (O(1) if tail is maintained)


```python
def insert_at_end(head, data):
    new_node = Node(data)
    if not head:
        return new_node

    temp = head
    while temp.next:
        temp = temp.next

    temp.next = new_node
    new_node.prev = temp
    return head
```


##🔸 Insert at Specific Position
- Insertion at a specific location in a Doubly Linked List (DLL) means adding a new node at a given position (index) by properly updating both prev and next pointers.

#### Algorithm:

- Create a new node with given data.

- If position == 0:

    - Insert at the beginning.

- Traverse the list to reach the node at (position − 1).

- Set newNode.next to current node’s next.

- Set newNode.prev to current node.

- Update current.next.prev to newNode (if it exists).

- Set current.next = newNode.


```python
def insert_at_position(head, data, pos):
    new_node = Node(data)
    temp = head

    for _ in range(pos - 1):
        temp = temp.next

    new_node.next = temp.next
    new_node.prev = temp
    if temp.next:
        temp.next.prev = new_node
    temp.next = new_node

    return head
```

## 🔸 Delete from Beginning
- Deletion from the beginning means removing the first node (head) of the doubly linked list.

#### Algorithm:

- If head == NULL, list is empty → stop.

- Move head to head.next.

- Set head.prev = NULL (if list is not empty).

- Old first node is removed.

#### Time Complexity: O(1)


```python
def delete_from_beginning(head):
    if not head:
        return None
    head = head.next
    if head:
        head.prev = None
    return head
```

```python
DeleteFromBeginning(head):
    if head == NULL:
        return NULL

    head = head.next
    if head != NULL:
        head.prev = NULL

    return head
```


##🔸 Delete from End
- Deletion from the end means removing the last node (tail) of the doubly linked list.

#### Algorithm:

- If head == NULL, list is empty → stop.

- Traverse to the last node.

- Set last.prev.next = NULL.

- Remove last node.

####Time Complexity: O(n)



```python
def delete_from_end(head):
    if not head:
        return None

    temp = head
    while temp.next:
        temp = temp.next

    if temp.prev:
        temp.prev.next = None
    else:
        head = None

    return head
```

```python
DeleteFromEnd(head):
    if head == NULL:
        return NULL

    temp = head
    while temp.next != NULL:
        temp = temp.next

    if temp.prev != NULL:
        temp.prev.next = NULL
    else:
        head = NULL

    return head
```

##🔸 Delete from Specific Position
- Deletion at a specific location means removing a node at a given position/index from the doubly linked list.

#### Algorithm:

- If head == NULL, list is empty → stop.

- Traverse to the node at the given position.

- Update:

    - node.prev.next = node.next

    - node.next.prev = node.prev (if exists)

- Remove the node.

####Time Complexity: O(n)


```python
def delete_at_position(head, pos):
    temp = head

    for _ in range(pos):
        temp = temp.next

    temp.prev.next = temp.next
    if temp.next:
        temp.next.prev = temp.prev

    return head
```

##🔹 Advantages of Doubly Linked List

- Bidirectional traversal
- Easy deletion (no need to track previous manually)
- Efficient undo/redo operations
- Used in navigation systems


<br>

##🔹 Disadvantages of Doubly Linked List

- Extra memory (extra pointer)
- More complex logic
- Slightly slower due to overhead
- Not cache-friendly

<br>

##🔹 Applications of Doubly Linked List

- Browser Back / Forward

- Undo / Redo in editors

- Music playlist navigation

- Implementing Deque

- LRU Cache
